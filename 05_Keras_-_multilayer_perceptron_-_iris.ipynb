```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.utils import to_categorical
from sklearn import datasets
```

```python
iris = datasets.load_iris()
X = iris.data
target = iris.target
Y = to_categorical(target).astype("uint8")
# 0 => [1 0 0]
# 1 => [0 1 0]
# 2 => [0 0 1]
```

```python
idx = 0
print(X[idx])
print(target[idx])
print(Y[idx])
```

## Corrida de referencia 4 × 3 × 3 (Keras)

```python
tf.keras.utils.set_random_seed(42)

model_ref = keras.Sequential(
  [
    layers.Dense(3, activation="sigmoid", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="sigmoid", name="layer2"),
  ]
)

model_ref.summary()
```

```python
optimizer = tf.keras.optimizers.SGD(learning_rate=0.03)
loss = tf.keras.losses.MeanSquaredError()

model_ref.compile(optimizer, loss)
history_ref = model_ref.fit(X, Y, epochs=500, verbose=0)
acc_ref = (np.argmax(model_ref.predict(X, verbose=0), axis=1) == target).mean()
print(f"Referencia 4x3x3 -> loss final: {history_ref.history['loss'][-1]:.4f} | precisión: {acc_ref:.3f}")

plt.plot(history_ref.history['loss'])
plt.title('4 x 3 x 3 (referencia)')
plt.show()
```

## Ejercicio 1 — Topología más profunda: 4 × 3 × 3 × 3 × 3 (Keras)
Se añaden **dos** `Dense(3, activation="sigmoid")` antes de la capa de salida. `model.summary()` debe mostrar 4 capas `Dense`. Mismo optimizador (SGD, η = 0.03), misma pérdida y 500 épocas.

```python
tf.keras.utils.set_random_seed(42)

model_deep = keras.Sequential(
  [
    layers.Dense(3, activation="sigmoid", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="sigmoid", name="layer2"),   # nueva (oculta 2)
    layers.Dense(3, activation="sigmoid", name="layer3"),   # nueva (oculta 3)
    layers.Dense(3, activation="sigmoid", name="layer4"),   # salida
  ]
)

model_deep.summary()
```

```python
optimizer = tf.keras.optimizers.SGD(learning_rate=0.03)
loss = tf.keras.losses.MeanSquaredError()

model_deep.compile(optimizer, loss)
history_deep = model_deep.fit(X, Y, epochs=500, verbose=0)
acc_deep = (np.argmax(model_deep.predict(X, verbose=0), axis=1) == target).mean()
print(f"Profunda 4x3x3x3x3 -> loss final: {history_deep.history['loss'][-1]:.4f} | precisión: {acc_deep:.3f}")

plt.plot(history_deep.history['loss'])
plt.title('4 x 3 x 3 x 3 x 3')
plt.show()
```

## Comparación: original vs. más profunda (Keras)

```python
plt.plot(history_ref.history['loss'],  label="4 x 3 x 3 (referencia)")
plt.plot(history_deep.history['loss'], label="4 x 3 x 3 x 3 x 3")
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()

print("Topología            | Loss inicial | Loss final | Precisión")
print(f"4 x 3 x 3            | {history_ref.history['loss'][0]:.4f}       | {history_ref.history['loss'][-1]:.4f}     | {acc_ref:.3f}")
print(f"4 x 3 x 3 x 3 x 3    | {history_deep.history['loss'][0]:.4f}       | {history_deep.history['loss'][-1]:.4f}     | {acc_deep:.3f}")
```
